```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
# 2. Inspection and Properties

print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# 3. Array Operations

# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Fahrenheit temperatures:", temperatures_f)

# Maximum, minimum and mean
print("Maximum temperature:", temperatures_w1.max())
print("Minimum temperature:", temperatures_w1.min())
print("Mean temperature:", temperatures_w1.mean())
```

```python
# 4. Array Slicing and Indexing

# First three days
print("First three days:", temperatures_w1[:3])

# Weekend (last two days)
print("Weekend:", temperatures_w1[-2:])

# Middle three days
print("Middle three days:", temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures:")
print(temperatures)
```

```python
# 6. Inspect the 2D Array

print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

# Week 1
print("Week 1:", temperatures[0])

# Week 2
print("Week 2:", temperatures[1])

# Weekend of Week 1 (last two days)
print("Week 1 weekend:", temperatures[0, -2:])

# Weekend of Week 2 (last two days)
print("Week 2 weekend:", temperatures[1, -2:])
```

```python
import pandas as pd

# 1. Creating Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
# 2. Indexing and Slicing

# Mark of 1st rank student
print("1st rank:", marks.iloc[0])

# Top 3 ranks using loc
print("Top 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Mark of 3rd rank using iloc
print("3rd rank:", marks.iloc[2])

# Marks greater than 90
print("Marks greater than 90:")
print(marks[marks > 90])
```

```python
# 3. Manipulating Series

# Modify 1st rank to 100
marks['Rank1'] = 100
print("After modifying Rank1:")
print(marks)

# Remove last rank
marks = marks.drop('Rank5')
print("After removing Rank5:")
print(marks)

# Calculate CGPA
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
import pandas as pd

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Furniture", "Clothing",
        "Furniture", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Display DataFrame
print(transactions)

# Head
print(transactions.head())

# Tail
print(transactions.tail())

# Shape
print(transactions.shape)

# Column names
print(transactions.columns)

# Data types
print(transactions.dtypes)

# ProductCategory and Amount
print(transactions[["ProductCategory", "Amount"]])

# Last 3 columns
print(transactions.iloc[:, -3:])

# Region is North AND Amount > 200
print(transactions[
    (transactions["Region"] == "North") &
    (transactions["Amount"] > 200)
])

# Value counts of ProductCategory
print(transactions["ProductCategory"].value_counts())

# Unique regions
print(transactions["Region"].unique())

# Mean Amount for each Region
print(transactions.groupby("Region")["Amount"].mean())
```

```python
# 1. Modify Amount for TransactionID 102 to 165
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

# 2. Add Discount column (10% of Amount)
transactions["Discount"] = transactions["Amount"] * 0.10

# 3. Remove the row with TransactionID 109
transactions = transactions[transactions["TransactionID"] != 109]

# 4. Delete the Discount column
transactions = transactions.drop("Discount", axis=1)

# Display final DataFrame
print(transactions)
```
